# Open CSV File

```python
import pandas as pd

# Replace 'your_file.csv' with the actual path to your CSV file
try:
    df = pd.read_csv('air_quality_historical.csv')
    print("CSV file loaded successfully!")
    # Display the first 5 rows of the DataFrame
    display(df.head())
except FileNotFoundError:
    print("Error: 'your_file.csv' not found. Please make sure the file exists and the path is correct.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
# Calculate mean, median, and mode for 'pm10'
pm10_mean = df['pm10'].mean()
pm10_median = df['pm10'].median()
pm10_mode = df['pm10'].mode()

print(f"Mean of PM10: {pm10_mean:.2f}")
print(f"Median of PM10: {pm10_median:.2f}")
print(f"Mode of PM10: {pm10_mode.tolist()}") # .mode() can return multiple values, so convert to list
```
